Create XOR dataset + train a single-layer sigmoid model

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

tf.random.set_seed(42)
np.random.seed(42)

# XOR inputs (x1, x2)
X = np.array([
    [0., 0.],
    [0., 1.],
    [1., 0.],
    [1., 1.]
], dtype=np.float32)

# XOR outputs
y = np.array([
    [0.],
    [1.],
    [1.],
    [0.]
], dtype=np.float32)

print("X:\n", X)
print("y:\n", y)

# Baseline: single-layer (linear classifier with sigmoid)
model_linear = Sequential([
    Dense(1, activation="sigmoid", input_shape=(2,))
])

model_linear.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

history_linear = model_linear.fit(
    X, y,
    epochs=500,
    batch_size=4,
    verbose=0
)

loss_lin, acc_lin = model_linear.evaluate(X, y, verbose=0)
print("Linear XOR Test loss:", loss_lin)
print("Linear XOR Test accuracy:", acc_lin)

# Predictions
pred_lin = model_linear.predict(X, verbose=0)
print("Predictions (probabilities):\n", np.round(pred_lin, 3))
print("Predicted classes:\n", (pred_lin >= 0.5).astype(int).reshape(-1))
```

Solve XOR using a Multi-Layer Perceptron (hidden layer + ReLU)

```python
# MLP for XOR (non-linear model)
model_mlp = Sequential([
    Dense(8, activation="relu", input_shape=(2,)),
    Dense(1, activation="sigmoid")
])

model_mlp.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

history_mlp = model_mlp.fit(
    X, y,
    epochs=1000,
    batch_size=4,
    verbose=0
)

loss_mlp, acc_mlp = model_mlp.evaluate(X, y, verbose=0)
print("MLP XOR Test loss:", loss_mlp)
print("MLP XOR Test accuracy:", acc_mlp)

# Predictions
pred_mlp = model_mlp.predict(X, verbose=0)
print("Predictions (probabilities):\n", np.round(pred_mlp, 3))
print("Predicted classes:\n", (pred_mlp >= 0.5).astype(int).reshape(-1))
```
